### 2. Handling Outliers

````python
# Detect outliers using the IQR method

numeric_columns = ['Open', 'High', 'Low', 'Close']

outlier_summary = {}

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary[col] = len(outliers)

print("Number of outliers detected using IQR:")
for col, count in outlier_summary.items():
    print(f"{col}: {count}")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The IQR (Interquartile Range) method was used to detect potential outliers in the numerical stock-price variables. The analysis identified 9 potential outliers each in the Open, High, Low and Close variables.

The detected observations were not removed or capped because extreme stock-price values can represent genuine market movements and important historical events rather than data-entry errors. Removing such observations could lead to loss of useful information and distort the historical behaviour of the stock. Therefore, the observations were retained in the dataset, while the IQR method was used for identification and monitoring of extreme values.Answer Here.

### 4. Feature Manipulation & Selection

#### 1. Feature Manipulation

````python
# Feature Manipulation

# Convert Date into datetime
df['Date'] = pd.to_datetime(
    df['Date'].astype(str).str.strip(),
    format='%b-%y',
    errors='coerce'
)

# Sort chronologically
df = df.sort_values('Date').reset_index(drop=True)

# Create engineered features
df['Price_Range'] = df['High'] - df['Low']
df['Open_Close_Change'] = df['Close'] - df['Open']
df['Monthly_Return'] = df['Close'].pct_change() * 100
df['Volatility'] = ((df['High'] - df['Low']) / df['Open']) * 100

# Create time-based features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Target: next month's closing price
df['Target_Close'] = df['Close'].shift(-1)

# Remove rows created by percentage change / future target
df = df.dropna().reset_index(drop=True)

print("Feature engineering completed.")
print("Dataset shape:", df.shape)

df.head()
````

#### 2. Feature Selection

````python
# Feature Selection

# Candidate features
features = [
    'Open',
    'High',
    'Low',
    'Close',
    'Price_Range',
    'Open_Close_Change',
    'Monthly_Return',
    'Volatility',
    'Year',
    'Month'
]

# Correlation of features with the target
correlation_with_target = (
    df[features + ['Target_Close']]
    .corr()['Target_Close']
    .drop('Target_Close')
    .sort_values(ascending=False)
)

print("Correlation of features with Target_Close:")
print(correlation_with_target)
````

##### What all feature selection methods have you used  and why?

Correlation analysis and domain-based feature selection were used to select relevant variables for the regression model. Correlation with the target variable, Target_Close, was examined to understand the strength and direction of the relationship between each candidate feature and the target. Domain relevance was also considered because Open, High, Low and Close prices contain important information about stock-price behaviour. Features with useful predictive information were retained while unnecessary or highly redundant features were avoided to reduce the risk of overfitting and multicollinearity.Answer Here.

##### Which all features you found important and why?

The important features considered for predicting Target_Close are Open, High, Low, Close, Price_Range, Open_Close_Change, Monthly_Return, Volatility, Year and Month. The current Open, High, Low and Close prices provide direct information about recent stock-price behaviour, while Price_Range, Open_Close_Change, Monthly_Return and Volatility capture additional price movement and risk characteristics. ## ***7. ML Model Implementation***

### ML Model - 1

````python
# ML Model 1 - Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt

# Create Linear Regression pipeline with StandardScaler
linear_model = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LinearRegression())
])

# Fit the model
linear_model.fit(X_train, y_train)

# Predict on test data
y_pred_lr = linear_model.predict(X_test)

# Evaluation metrics
mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression Performance")
print("--------------------------------")
print("MAE :", mae_lr)
print("MSE :", mse_lr)
print("RMSE:", rmse_lr)
print("R²  :", r2_lr)
````

Linear Regression was used as the first machine learning model because it provides a simple and interpretable baseline for predicting the next month's closing price. The model uses the selected stock-price and engineered features to estimate Target_Close.

StandardScaler was incorporated into the model pipeline so that features with different numerical ranges are standardized without causing data leakage. The model was trained on the chronological training dataset and evaluated on the later unseen test dataset.

The model performance is evaluated using Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE) and R² score. Lower MAE and RMSE indicate smaller prediction errors, while a higher R² indicates that the model explains a larger proportion of the variation in the target variable.

````python
# Evaluation Metric Score Chart

metrics = ['MAE', 'RMSE', 'R²']
values = [mae_lr, rmse_lr, r2_lr]

plt.figure(figsize=(8, 5))
plt.bar(metrics, values)
plt.title('Linear Regression - Evaluation Metrics')
plt.ylabel('Score')
plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Cross-Validation & Hyperparameter Tuning
# Ridge Regression with TimeSeriesSplit and GridSearchCV

from sklearn.linear_model import Ridge
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create pipeline
ridge_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge())
])

# Hyperparameter values
param_grid = {
    'model__alpha': [0.01, 0.1, 1, 10, 100]
}

# Time-series cross-validation
tscv = TimeSeriesSplit(n_splits=5)

# Grid Search
grid_search = GridSearchCV(
    estimator=ridge_pipeline,
    param_grid=param_grid,
    cv=tscv,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

# Fit
grid_search.fit(X_train, y_train)

# Best parameters
print("Best Hyperparameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation RMSE:")
print(np.sqrt(-grid_search.best_score_))

# Best model
best_ridge_model = grid_search.best_estimator_

# Prediction on test data
y_pred_ridge = best_ridge_model.predict(X_test)

# Evaluation
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mse_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)

print("\nTuned Ridge Regression Performance")
print("-----------------------------------")
print("MAE :", mae_ridge)
print("MSE :", mse_ridge)
print("RMSE:", rmse_ridge)
print("R²  :", r2_ridge)
````

##### Which hyperparameter optimization technique have you used and why?

Grid Search Cross-Validation was used for hyperparameter optimization. TimeSeriesSplit with 5 splits was used instead of random cross-validation because the dataset contains chronological stock-price observations. The Ridge Regression model was tested with different values of the alpha parameter: 0.01, 0.1, 1, 10 and 100. The best parameter selected by GridSearchCV was alpha = 1. This approach helps identify a suitable regularization strength while evaluating the model using multiple chronological training and validation splits.Answer Here.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

The tuned Ridge Regression model showed mixed improvement compared with the original Linear Regression model. The R² score improved substantially, reaching approximately 0.896, indicating that the tuned model explains a large proportion of the variation in the target variable. However, the MAE and RMSE did not decrease compared with the original model. Therefore, the tuning improved the model's explanatory power, but it did not improve all prediction-error metrics. Further comparison with nonlinear models is required before selecting the final model.Answer Here.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Random Forest - Cross-Validation & Hyperparameter Tuning

from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Random Forest model
rf_base = RandomForestRegressor(
    random_state=42
)

# Hyperparameter grid
rf_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10, 15],
    'min_samples_split': [2, 5]
}

# Time-series cross-validation
tscv_rf = TimeSeriesSplit(n_splits=5)

# Grid Search
rf_grid_search = GridSearchCV(
    estimator=rf_base,
    param_grid=rf_param_grid,
    cv=tscv_rf,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

# Fit
rf_grid_search.fit(X_train, y_train)

# Best parameters
print("Best Random Forest Hyperparameters:")
print(rf_grid_search.best_params_)

print("\nBest Cross-Validation RMSE:")
print(np.sqrt(-rf_grid_search.best_score_))

# Best Random Forest model
best_rf_model = rf_grid_search.best_estimator_

# Predict on test data
y_pred_rf_tuned = best_rf_model.predict(X_test)

# Evaluation
mae_rf_tuned = mean_absolute_error(y_test, y_pred_rf_tuned)
mse_rf_tuned = mean_squared_error(y_test, y_pred_rf_tuned)
rmse_rf_tuned = np.sqrt(mse_rf_tuned)
r2_rf_tuned = r2_score(y_test, y_pred_rf_tuned)

print("\nTuned Random Forest Performance")
print("--------------------------------")
print("MAE :", mae_rf_tuned)
print("MSE :", mse_rf_tuned)
print("RMSE:", rmse_rf_tuned)
print("R²  :", r2_rf_tuned)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import GradientBoostingRegressor

# Gradient Boosting model
gb_base = GradientBoostingRegressor(random_state=42)

# Hyperparameter search space
param_grid_gb = {
    'n_estimators': [50, 100, 150, 200],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'max_depth': [2, 3, 4, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Randomized Search with Cross-Validation
random_search_gb = RandomizedSearchCV(
    estimator=gb_base,
    param_distributions=param_grid_gb,
    n_iter=20,
    cv=5,
    scoring='neg_root_mean_squared_error',
    random_state=42,
    n_jobs=-1
)

# Fit the algorithm
random_search_gb.fit(X_train, y_train)

# Best model
best_gb = random_search_gb.best_estimator_

# Predict on the test data
y_pred_gb_tuned = best_gb.predict(X_test)

# Evaluation
mae_gb_tuned = mean_absolute_error(y_test, y_pred_gb_tuned)
mse_gb_tuned = mean_squared_error(y_test, y_pred_gb_tuned)
rmse_gb_tuned = np.sqrt(mse_gb_tuned)
r2_gb_tuned = r2_score(y_test, y_pred_gb_tuned)

print("Best Gradient Boosting Parameters:")
print(random_search_gb.best_params_)

print("\nBest Cross-Validation RMSE:")
print(-random_search_gb.best_score_)

print("\nTuned Gradient Boosting Performance")
print("------------------------------------")
print("MAE :", mae_gb_tuned)
print("MSE :", mse_gb_tuned)
print("RMSE:", rmse_gb_tuned)
print("R²  :", r2_gb_tuned)
````
